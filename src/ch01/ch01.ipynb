## この Notebook で確認すること

- 現在の `uv` 環境で Python と主要パッケージのバージョンを確認する。
- Matplotlib がヘッドレス実行環境でも動作することを確認する。
- 原本リポジトリの読み取り専用コンテンツから章の図版を読み込めることを確認する。

```python
from importlib.metadata import version
from pathlib import Path
import platform
import sys

import matplotlib
import matplotlib.pyplot as plt
import pandas as pd

PACKAGE_NAMES = ["numpy", "pandas", "matplotlib", "scikit-learn", "torch"]

print(f"Python 実行ファイル: {sys.executable}")
print(f"Python バージョン: {platform.python_version()}")
print(f"Matplotlib バックエンド: {matplotlib.get_backend()}")
```

```python
package_versions = pd.DataFrame(
    [(name, version(name)) for name in PACKAGE_NAMES],
    columns=["パッケージ", "バージョン"],
)
package_versions
```

## ヘッドレス描画の確認

CI ワークフローでは `MPLBACKEND=Agg` を設定しているため、`plt.show()` を呼び出しても Notebook 実行をブロックしないことを確認します。

```python
fig, ax = plt.subplots(figsize=(4, 2.5))
ax.plot([0, 1, 2, 3], [0.1, 0.4, 0.35, 0.8], marker="o", linewidth=2)
ax.set_title("ヘッドレス描画の確認")
ax.set_xlabel("ステップ")
ax.set_ylabel("スコア")
ax.grid(True, alpha=0.3)
plt.show()
plt.close(fig)
```

## 原本リポジトリの図版を再利用する

移行後の Notebook は `src/` 配下に配置しますが、書籍の正本となる図版アセットは読み取り専用の `machine-learning-book/` サブモジュールに残します。

```python
from IPython.display import Image, display

fig_dir = None
for candidate in [Path.cwd(), *Path.cwd().parents]:
    possible = candidate / "machine-learning-book/ch01/figures"
    if possible.exists():
        fig_dir = possible.resolve()
        break

assert fig_dir is not None, "現在の作業ディレクトリから machine-learning-book/ch01/figures を見つけられませんでした"

selected_figures = [
    fig_dir / "01_01.png",
    fig_dir / "01_06.png",
    fig_dir / "01_09.png",
]

for figure_path in selected_figures:
    print(figure_path.name)
    display(Image(filename=str(figure_path), width=600))
```
